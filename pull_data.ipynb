```python
import yfinance as yf
import pandas as pd

crypto_tickers = {
    "Bitcoin":      "BTC-USD",
    "Ethereum":     "ETH-USD",
    "XRP":          "XRP-USD",
    "Litecoin":     "LTC-USD",
    "Polkadot":     "DOT-USD",
    "Bitcoin Cash": "BCH-USD",
    "Binance Coin": "BNB-USD",
    "Cardano":      "ADA-USD",
    "Chainlink":    "LINK-USD",
    "Bitcoin SV":   "BSV-USD",
    "Stellar":      "XLM-USD",
    "EOS":          "EOS-USD",
    "Monero":       "XMR-USD",
    "TRON":         "TRX-USD",
    "Tezos":        "XTZ-USD",
    "Uniswap":      "UNI-USD",
    "VeChain":      "VET-USD",
    "Cosmos":       "ATOM-USD",
    "Aave":         "AAVE-USD",
    "Neo":          "NEO-USD"
}

all_prices = {}

for name, ticker in crypto_tickers.items():
    try:
        df = yf.download(ticker, start="2021-01-01", end="2026-01-01",
                         interval="1d", progress=False, auto_adjust=True)
        if df.empty:
            print(f"{name} ({ticker}): no data")
            continue
        if isinstance(df.columns, pd.MultiIndex):
            df.columns = df.columns.get_level_values(0)
        all_prices[name] = df["Close"]
        print(f"{name} ({ticker}): {len(df)} days pulled")
    except Exception as e:
        print(f"{name} ({ticker}): error — {e}")

crypto_prices = pd.DataFrame(all_prices)
crypto_prices.index.name = "Date"

print(f"\nShape: {crypto_prices.shape}")
print(f"\nMissing values:\n{crypto_prices.isnull().sum()}")
print(f"\nPreview:\n{crypto_prices.head()}")

crypto_prices.to_excel("crypto_daily_prices.xlsx")
print("\nSaved to crypto_daily_prices.xlsx")
```
